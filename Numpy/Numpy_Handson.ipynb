```python
# Take 5 random integers from 0 to 100 in an array
# Find the mean of the array
# Replace values below mean with 0
# Count how many values were replaced
```

```python
np.random.seed(0)
arr = np.random.randint(0,101,size=10)
print(arr)
mean = np.mean(arr)
print(mean)
count = np.sum(arr < mean)
print(count)
arr[arr < mean] = 0
print(arr)
```
