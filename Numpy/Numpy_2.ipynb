```python
import numpy as np
```

```python
# sum

a = np.array([1,2,3,4,5,6])
ans = np.sum(a)
print(ans)
```

```python
# Sorting

a = np.array([21,2,3,4,7,91,56,78,45,100,8])
```

```python
arr = np.sort(a)
```

```python
arr
```

```python
a = np.array([21,2,3,4,7,91,56,78,45,100,8])
arr_new = np.sort(a)[::-3] # start index : stop index : step size
arr_new
```

```python
# Mean

a = np.array([3,1,4,5,6,7,2,9])
a
```

```python
# Median - middle value

a = np.array([3,1,4,5,6,7,2,9,10])
a
np.sort(a)
```

```python
med = np.median(a)
med
```

```python
# Mode - Value with higher frequency

a = np.array([3,1,4,5,6,4,4,7,2,9,10])
m = np.mode(a)   # mode is defined with stats lib called scipy
m
```

```python
from scipy import stats
```

```python
a = np.array([3,1,4,5,6,4,4,7,2,9,10])
m = stats.mode(a)   # mode is defined with stats lib called scipy
print(m.mode)
```

```python
import statistics

m = statistics.mode(a)
print(m)
```

```python
m1 = np.array([[1,2,3],[4,5,6]])
print(m1)
```

```python
# rows - 2 , axis = 1
# columns = 3 , axis = 0
```

```python
m1_f = m1.flatten()
print(m1_f)
```

```python
# Higher dimensional array -> 1D
```

```python
m1_f[2] = 30
m1_f
```

```python
print(m1)
```

```python
# Ravel()

# Convert the matrix to 1D

# Changes in the ravel array will directly affect my original array
```

```python
m1_r = m1.ravel()
m1_r
```

```python
m1_r[3] = 40
m1_r
```

```python
print(m1)
```
