```python
#VARIANCE : Variance measures how far all data points are from the mean.
#-> checks every value
# -> compares each value with average
# -> measures overall spread properly


#High variance - more spread out data
#Low variance - less spread out data -> data points are closer to the mean


#Class A marks: 10,20,50,80,90
#Class B marks: 48,49,50,51,52

# -> Both classes may have similar avg
# -> But class A has much larger spread
# -> Variance detects this inconsistency , which range alone cannot explain properly


data=[50,70,60,90,35]
mean=np.mean(data)
print(mean)
var= ((50-61)**2+(70-61)**2+(60-61)**2+(90-61)**2+(35-61)**2)/5
print(var)
```

```python
#OUTLIERS : An outlier is a data point that is very different from the rest of the data points
#  -> extremely high or extremely low values present in our data are known as outliers
#  -> outliers can affect our data analysis
```

```python
sal = [30000,45000,40000,28000,35000,55000,60000,72000,300000,800000]
print(np.mean(sal))
```

```python
# Visual representation of outliers
import pandas as pd
import statistics as stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(sal)
plt.show()
```

```python
# How do we remove the outliers?

# -> IQR : Inter-Quartile Range

# 1- Find the Q1 (25th percentile) and Q3 (75th percentile)
# 2- IQR = Q3 - Q1
# 3- LB(Lower Bound) = Q1 - 1.5 * IQR
# 4- UB(Upper Bound) = Q3 + 1.5 * IQR

# -> Any value outside the LB or UB is considered as outlier
```

```python
a1=[-100,-45,-18,25,30,467,45,68,76,67,98,250,76,99,271]
q1=np.percentile(a1,25)
q3=np.percentile(a1,75)
print(q1)
print(q3)
```

```python
IQR = q3-q1
print(IQR)
```

```python
LB= q1-1.5*IQR
print(LB)
```

```python
UB=q3+1.5*IQR
print(UB)
```

```python
clean_data=[]
for i in a1:
  if i<LB or i>UB:
    print(f'The outlier in the data is {i} and it has been removed')
  else:
    clean_data.append(i)
```

```python
clean_data
```

```python
for i in a1.copy():
  if i<LB or i>UB:
    a1.remove(i)

print(a1)
```

```python
#We need distribution :-
# to know where our most values lie
# whether data is balanced
# whether data is symmetric or skewed
```

```python
#NORMAL DISTRIBUTION :
# A normal distribution is a distribution where :
# most values are near the average
# Extreme values are rare
# shape looks like a bell

#Key characteristics:
#Mean = Mode = Median
#Symmetric shape
#Bell shaped curve
```

```python
x1=[15,16,16,17,17,17,17,18,18,19]
print(np.mean(x1))
plt.hist(x1,bins=5,color='green',edgecolor='red')
plt.title('Normal dist.')
plt.show()
```

```python
#SKEWNESS : which side the data is stretched towards

#Types of skewness :

# 1- Positive skewness (Right skew)
# Long tail on the right side
# Most values are small , few values are extremely large
# Ex: salaries - most earn moderate , few extremely high

from scipy.stats import skew

x=[10,10,10,10,10,10,10,11,11,12,13]
print(np.mean(x))
print(np.median(x))
m=stats.mode(x)
print(m)
print(skew(x))
plt.hist(x,bins=4,color='green',edgecolor='red')
plt.title('Right Skew')
plt.show()
```

```python
# Negative skewness (Left skew):
# Long tail on left side
# Most values are high , few are small
# Ex- Easy exam : most stud score high , few will fail badly

x2=[10,11,11,12,12,12,12,12]
print(skew(x2))
print(np.mean(x2))
print(np.median(x2))
m=stats.mode(x2)
print(m)
plt.hist(x2,bins=3,color='skyblue',edgecolor='red')
plt.title('Left Skewed')
plt.show()
```

```python
#Mean and skewness relation:
# +ve skew : Mode<Median<Mean
# -ve skew : Mean<Median<Mode


#We often interpret skewness
# 0  Symmetric
# -0.5 to 0.5   Approximately Symmetric
# -1 to -0.5 or 0.5 to 1   Moderateky skewed
# Less than -1 or greater than +1   Highly skewed
```
