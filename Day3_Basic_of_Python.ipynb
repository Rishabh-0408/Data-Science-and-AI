Q.Verify the Formula Divident = quotient*divisor+reminder

```python
divisor = int(input())
divident = int(input())
quotient = divident // divisor
reminder = divident % divisor
verify = quotient*divisor+reminder
print("quotient",quotient)
print("reminder:",reminder)
print("verified:",verify)
```

Q. Convert secs into Hours and minitues

```python
seconds = int(input())
minutes = seconds // 60
hours = minutes // 60
remaining_minutes = minutes % 60
remaining_seconds = seconds % 60
print(f"hours:minutes:sec = {hours}:{remaining_minutes}:{remaining_seconds}")
```

Q. convert days into years and month

```python
days = int(input())
year = days // 365
remaining_days = days % 365
days = remaining_days % 30
month =  remaining_days // 30
print(f"years:months:days = {year}:{month}:{days}")
```

**Comparison Operator**

```python
a = int(input())
b = int(input())
print(f"a > b: {a > b}")
print(f"a < b: {a < b}")
print(f"a == b: {a == b}")
print(f"a != b: {a != b}")
print(f"a >= b: {a >= b}")
print(f"a <= b: {a <= b}")
```

**Assignment Operator**

```python
a = 30
print(a)
a+=30
print("a+=30:",a)
a-=20
print("a-=20:",a)
a*=2
print("a*=2:",a)
a//=2
print("a//=2:",a)
```

**Logical Operator :** and, or, not

```python
print(10>5 and 15<10)
```

```python
a = 10
b = 15
print(a==b or a<b)
```

**Bitwise Operator :** and,or,xor

```python
a = 7
b = 19
print(f"a&b: {a&b}")
print(f"a|b: {a|b}")
print(f"a^b: {a^b}")
print(f"~a: {~a}")
print(f"~b: {~b}")
#Left Shift
print(f"a<<1: {a<<1}")
#Right shift
print(f"a>>1: {a>>1}")
```
