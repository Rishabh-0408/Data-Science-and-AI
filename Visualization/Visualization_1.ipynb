```python
days = ["mon","tue","wed","thur","fri","sat","sun"]
sales = [20,23,250,26,145,123,89]
plt.plot(days, sales, marker = 'o', color = 'blue', linestyle = "dashed", markersize = 10, markerfacecolor = "red", markeredgecolor = "yellow")
plt.grid(True)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Plot for days and sales of the product")
#we want the the data of every day we use loop
for i in range (len(days)):
  plt.text(days[i],sales[i]+20, str(sales[i]), ha = "center", va = "bottom")
  #ha -> horizontal allignment
  #va -> vertical allignment

plt.ylim(0,400)
plt.show()
```

```python
# Create a line plot to compare the sales data of three fruits in a week
days = ["mon","tue","wed","thur","fri","sat","sun"]
sales_Mango = [200,203,250,206,145,123,199]
sales_Apple = [200,100,400,250,300,500,150]
sales_Grape = [360,250,300,400,203,250,255]
plt.plot(days ,sales_Apple,marker = 'o', color = 'blue', linestyle = "dashed",markersize = 10, markerfacecolor = "yellow", markeredgecolor = "black")
plt.plot(days, sales_Grape,marker = 'o', color = 'red', linestyle = "dashed",markersize = 10, markerfacecolor = "lightblue", markeredgecolor = "black")
plt.plot(days, sales_Mango,marker = 'o', color = 'green', linestyle = "dashed",markersize = 10, markerfacecolor = "orange", markeredgecolor = "black")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Days vs Sales of Fruits")
plt.grid(True)
plt.ylim(0,600)
for i in range (len(days)):
  plt.text(days[i],sales_Mango[i]+20, str(sales_Mango[i]), ha = "center", va = "bottom")
  plt.text(days[i],sales_Grape[i]+20, str(sales_Grape[i]), ha = "center", va = "bottom")
  plt.text(days[i],sales_Apple[i]+20, str(sales_Apple[i]), ha = "center", va = "bottom")
plt.show()
```

```python
# Create a bar plot of 8 Students and their marks in mathematics
# use bar labels
students = ['s1','s2','s3','s4','s5','s6','s7','s8']
Marks = [30,33,52,62,100,85,62,45]
bars = plt.bar(students, Marks, color = "red")
plt.bar_label(bars)
plt.grid(True)
plt.gca().set_facecolor("lightblue")
plt.xlabel("No of students")
plt.ylabel("Marks")
plt.title("Students vs Marks")
plt.show()
```

```python
# using Conditions in this plots
# students who have scored greater than 50 have green color other should have red color
students = ['s1','s2','s3','s4','s5','s6','s7','s8']
Marks = [30,33,52,62,100,85,62,45]
bars = plt.bar(students, Marks,color =["red" if m < 50 else "green" for m in Marks])
plt.bar_label(bars)
plt.grid(True)
plt.gca().set_facecolor("lightblue")
plt.xlabel("No of students")
plt.ylabel("Marks")
plt.title("Students vs Marks")
plt.show()
```

```python
student = ["Mark", "Toni", "Wiz", "Thomas", "Frank", "Sam","Salt", "Kevin"]
marks = [75, 62, 70, 95, 35, 69, 43, 83]
colors = ["green" if m >= 50 else "red" for m in marks]

bars = plt.bar(student, marks, color=colors)
plt.grid(True, color ="black")
plt.bar_label(bars)
plt.xlabel("student")
plt.ylabel("marks")
plt.title("Plot for students vs marks in maths")
plt.ylim(0,100)
plt.gca().set_facecolor("lightyellow")
plt.show()
```

```python
# Scatter plot
# Visulize the Spred of data
# Find out the relationship between numerical variable
rollno = [1,2,3,4,5,6,7,8]
marks = [75, 62, 70, 95, 35, 69, 43, 83]
plt.scatter(rollno,marks)
plt.xlabel("roolno")
plt.ylabel("marks")
plt.title("rollno vs marks")
plt.show()
```

```python
rollno = [1,2,3,4,5,6,7,8]
marks = [75, 62, 70, 95, 35, 69, 43, 83]
plt.scatter(rollno,marks, marker="*",color = "red",s = 200)
plt.xlabel("roolno")
plt.ylabel("marks")
plt.title("rollno vs marks")
plt.show()
```

```python
rollno = [1,2,3,4,5,6,7,8]
marks_maths = [75, 62, 70, 95, 35, 69, 43, 83]
marks_sci = [30,33,52,62,100,85,62,45]
plt.scatter(rollno,marks_maths,marker="*",color = "red",s = 100,label = "purple is for maths")
plt.scatter(rollno,marks_sci,marker="o",color = "purple",s = 100,label = "red is for science")
plt.xlabel("rollno")
plt.ylabel("marks")
plt.title("rollno vs marks")
plt.legend(fontsize=12, loc="best",frameon=True,edgecolor = "black")
plt.show()
```
