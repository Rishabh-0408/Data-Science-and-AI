```python
# Create a scatter plot between mpg vs hp with respect to brand
# create using both matplotlib and seaborn

# hue = data['brand']
```

```python
# using matplot
plt.scatter(df['mpg'],df['hp'],marker="*",s = 100,c = df['mpg'],cmap='viridis',edgecolors=(0, 0, 0, 0.5)) #s=markersize
plt.xlabel('MPG')
plt.ylabel('HP')
plt.title("MPG vs HP")
plt.gca().set_facecolor("white")
plt.show()
```

```python
#c=df['mpg'] — har point ka color uski mpg value ke hisaab se change hoga automatically.
# agar color use kar rahe ho to cmap/c use nahi kar sakte
#cmap -> Look
# 'viridis'  Purple → Yellow
# 'coolwarm' Blue → Red
# 'plasma' Purple → Orange
# 'rainbow' Full spectrum
```

```python
# using Seaborn
sns.scatterplot(
    data = df,
    x = 'mpg',
    y = 'hp',
    hue = 'brand',
)
plt.xlabel("Miles Per Gallon", fontsize=12, color='black', fontweight='bold')
plt.ylabel("Horsepower",       fontsize=12, color='black', fontweight='bold')
plt.title("MPG vs Horsepower by Brand (Seaborn)")
plt.legend(loc='best',bbox_to_anchor=(1.05, 1),frameon = True,edgecolor = 'black')
plt.tight_layout()
plt.show()
```

```python
#bbox_to_anchor=(1.05, 1) -> Legend ko plot ke bahar rakhhta haii
# (1.05, 1) → right side, top pe
# (0.5, -0.1) → neeche center mein
# (1.05, 0.5) → right side, middle mein
```

```python
# plt.tight_layout()
# Automatically spacing fix karta hai — legend, title, labels ek doosre ke upar na aaye.
```

```python
# Simple rule:
# Jab bhi legend bahar rakho bbox_to_anchor se → tight_layout() zaroor lagao, warna legend crop ho jaayega.
```

```python
fig =plt.subplots(2,1, figsize = (10,10))
plt.subplot(2,1,1)
sns.scatterplot(
    data = df,
    x = "mpg",
    y = "hp",
    hue = "brand"
)
plt.title("MPG vs HP")
plt.xlabel("MPG")
plt.ylabel("HP")

plt.subplot(2,1,2)
sns.scatterplot(
    data = df,
    x = "cylinders",
    y = "hp",
    hue = "brand"
)
plt.title("CYLINDERS vs HP")
plt.xlabel("CYLINDERS")
plt.ylabel("HP")
```

```python
# Mpg vs hp , Cylinders vs hp , HP vs MPG , HP vs Time-to-60
# Create sublots for the same
```

```python
fig =plt.subplots(2,2, figsize = (14,12))
plt.subplot(2,2,1)
sns.scatterplot(
    data = df,
    x = "mpg",
    y = "hp",
    hue = "brand"
)
plt.title("MPG vs HP")
plt.xlabel("MPG")
plt.ylabel("HP")

plt.subplot(2,2,2)
sns.scatterplot(
    data = df,
    x = "cylinders",
    y = "hp",
    hue = "brand"
)
plt.title("CYLINDERS vs HP")
plt.xlabel("CYLINDER")
plt.ylabel("HP")

plt.subplot(2,2,3)
sns.scatterplot(
    data = df,
    x = "hp",
    y = "mpg",
    hue = "brand"
)
plt.title("HP VS MPG")
plt.xlabel("HP")
plt.ylabel("MPG")

plt.subplot(2,2,4)
sns.scatterplot(
    data = df,
    x = "hp",
    y = "time-to-60",
    hue = "brand"
)
plt.title("HP VS TIME-TO-60")
plt.xlabel("HP")
plt.ylabel("TIME-TO-60")

plt.tight_layout()
plt.show()
```

```python
# Create  a bar plot for for cylinders vs hp with respect to brand
```

```python
sns.barplot(data = df, x = 'cylinders', y = 'hp', hue = 'brand')
plt.title("Cylinders vs hp with respect to brand",fontweight = "bold")
plt.xlabel("Cylinders", fontsize = 10, fontweight = "bold")
plt.ylabel("Hp", fontsize = 10, fontweight = "bold")
plt.legend(loc = "best",bbox_to_anchor=(1.02,0.8),frameon = True,edgecolor = 'black')
plt.tight_layout()
plt.show()
```

```python
# You have to create subplot
# two barplots in one row
# Cylinders vs hp , cylinders vs mpg
```

```python
fig = plt.subplots(2,1, figsize = (10,8))
plt.subplot(2,1,1)
sns.barplot(data = df, x = "cylinders", y = "hp", hue = "brand")
plt.title('Cylinders vs hp', fontweight = 'bold', fontsize = 20)
plt.xlabel('Cylinders', fontweight = 'bold', fontsize = 10)
plt.ylabel('hp', fontweight = 'bold', fontsize = 10)
plt.legend(loc = 'best', frameon = True, edgecolor = 'black', shadow = True)

plt.subplot(2,1,2)
sns.barplot(data = df, x = "cylinders", y = "mpg", hue = "brand")
plt.title('Cylinders vs mpg', fontweight = 'bold', fontsize = 20)
plt.xlabel('Cylinders', fontweight = 'bold', fontsize = 10)
plt.ylabel('mpg', fontweight = 'bold', fontsize = 10)
plt.legend(loc = 'best', frameon = True, edgecolor = 'black', shadow = True)

plt.tight_layout()
plt.show()
```
