```python
# which states are rarely contributing to honey production in last 27 years
```

```python
df.head()
```

```python
last_n_years = 27  # yahan 10, 20, 27 etc— any valid value

max_year = df['year'].max()           # 2021
min_year = max_year - last_n_years + 1 # 2012 (for 10 years)

df_filtered = df[df['year'] >= min_year]
state_prod = df_filtered.groupby('state')['production'].sum()
bottom_states = state_prod.sort_values().head(15).reset_index()
```

```python
bottom_states
```

```python
plt.figure(figsize=(20,8),facecolor='#FFFDD0')   # figure background
bars = sns.barplot(
    data = bottom_states,
    x = 'state',
    y = 'production',
    palette="YlOrRd",
    hue = 'state',
    linewidth=0,
)


# plt.gca().set_facecolor('#0f0f1a')           # plot area background
# plt.gca().tick_params(colors='white')         # tick labels white
# plt.gca().xaxis.label.set_color('white')     # x label white
# plt.gca().yaxis.label.set_color('white')     # y label white
# plt.gca().title.set_color('white')           # title white

plt.title('States are rarely contributing to honey production in last 27 years',fontweight = 'bold',fontsize = 20)
plt.xlabel("States")
plt.ylabel('Production')
plt.xticks(rotation = 45,ha='right')
plt.legend([], [], frameon=False)
plt.tight_layout()
plt.show()
```

```python
# Which was the year when the production of honey in whole US was Highest
```

```python
year_production = df.groupby('production')['year'].sum().reset_index()
highest_year_production = year_production.sort_values(by = 'production',ascending=False).head(10)
highest_year_production
```

```python
plt.plot(highest_year_production['production'],highest_year_production['year'],marker = 'o', markerfacecolor = 'lightblue', markeredgecolor = 'blue', color = 'black')
plt.title('Year when the production of honey in whole US was Highest')
plt.xlabel('Productions')
plt.ylabel('Years')
plt.show()
```

```python
# Which states have the highest number of colonies in the year 2000
```

```python
df.head(3)
```

```python
year2000 = df[df['year'] == 2000]
year2000 = year2000.sort_values(by = 'colonies_number', ascending=False).reset_index()
highest_state_colonies = year2000[['colonies_number','state']].head(10)
```

```python
highest_state_colonies
```

```python
plt.figure(figsize = (8,8))
plt.barh(highest_state_colonies['state'],highest_state_colonies['colonies_number'], color = sns.color_palette('coolwarm'))
plt.xticks(rotation = 0,ha = 'right')
plt.title("The highest number of colonies in the year 2000")
plt.xlabel('Colonies_number')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```
