```python
# Create a Bar plot showing number for games available in each genre
df.head()
```

```python
genre_totals = df['Genre'].value_counts().reset_index()
plt.figure(figsize = (15,8))
sns.barplot(x = 'Genre', y = 'count', data = genre_totals, palette = "Blues_d")
plt.xlabel('Genre')
plt.ylabel('Number of Games')
plt.title('Number of Games Available in Each Genre')
plt.xticks(rotation = 45)
plt.show()
```

```python
# Conclusion: Genres such as Action, Sports, and Misc dominate the gaming industry suggesting a strong consumer demand and need management's focus
# In contrast, Genres with fewer games may represent niche market or specialized player interest.
```

```python
# Find out Top platform by global_sales
df.head()
```

```python
platform_sales = df.groupby('Platform')['Global_Sales'].sum().reset_index()
platform_sales
```

```python
top_platfrom_sales = df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).reset_index()
top_platfrom_sales
```

```python
plt.figure(figsize=(12,6))
plt.bar(top_platfrom_sales['Platform'],top_platfrom_sales['Global_Sales'],edgecolor = 'black',color = 'skyblue')
plt.xticks(rotation = 45)
plt.xlabel('Platform')
plt.ylabel('Global Sales')
plt.title('Top 10 Platforms by Global Sales')
plt.show()
```

```python
# Conclusion: A small number of gaming platforms dominate global sales, such as PS2, X360, PS3, Wii, and DS, reflecting their strong market presence an Platforms
# with higher glonbal sales likely benefitted from popular game titles, better hardward adoption, arge user community
# and long-term customer loyatly.
```

```python
# Analyse the trend in Global sales over the years
```

```python
df.head(3)
```

```python
Sales_Trends = df.groupby('Year')['Global_Sales'].sum().sort_values(ascending = True).reset_index()
Sales_Trends
```

```python
# There was growth in the sales from 1980 to 1995 and it kept on increasing from the year 2000 and was highest around 2010 and started falling after 2014 and declined by 2020
```

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(10,8))
sns.lineplot(data = Sales_Trends, x = 'Year', y = 'Global_Sales',color = 'black',marker = "o", markeredgecolor = "Green",markerfacecolor = 'yellow', markersize = 7,linestyle = "dashed")
plt.title('Trend in Global sales over the years', fontweight = 'bold', fontsize = 20)
plt.xlabel('Years')
plt.ylabel('Global_Sales')
for i in range(len(Sales_Trends['Year'])):
  plt.text(
      Sales_Trends['Year'].iloc[i],
      Sales_Trends['Global_Sales'].iloc[i],
      # str(round(Sales_Trends['Global_Sales'].iloc[i],2)),
      # ha = 'center', va = 'bottom'
      f"{Sales_Trends['Global_Sales'].iloc[i]:.2f}",
      color = 'red'
  )
plt.grid(True , alpha = 0.8, linestyle = '--', axis = 'both')
plt.tight_layout()
plt.show()
```

```python
# There was growth in the sales from 1980 to 1995 and it kept on increasing from the year 2000 and
# was highest around 2010 and started falling after 2014 and declined by 2020
```

```python
# Compare and find out the relatioship between NA sales and EU sales
```

```python
df.head(2)
```

```python
plt.figure(figsize=(10,8))
sns.scatterplot(
    data = df,
    x = 'NA_Sales',
    y = 'EU_Sales',
    hue = 'Genre'
)
plt.title("NA Sales vs EU Sales by Genre", fontweight = "bold", fontsize = 15)
plt.xlabel("NA Sales")
plt.ylabel("EU Sales")
plt.legend(title = 'Genre', frameon = True, shadow = True, edgecolor = 'black', loc = 'best', bbox_to_anchor = (1,1))

plt.tight_layout()
plt.show()
```

```python
# #Conclusion: There is a positive relationship between NA and EU sales,
#  indicating that games popular in NA tend to perform well in Europe as well
#  perhaps because of similar tastes.
# However, the spread of points in scateer plot suggests that regional preferences and market dynamics
#  also influence sales differently across two regions.
```

```python
NA_group = df.groupby('Year')['NA_Sales'].sum().reset_index()
NA_group
```

```python
EU_group = df.groupby('Year')['EU_Sales'].sum().reset_index()
EU_group
```

```python
plt.figure(figsize = (15,8))
sns.lineplot(data = NA_group, x = 'Year', y = 'NA_Sales', marker = "o", label = "Blue is NA")
sns.lineplot(data = EU_group, x = 'Year', y = 'EU_Sales', marker = "o", label = "Orange is EU")
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('NA vs. EU Sales')
plt.legend()
plt.grid(True, axis = 'both', linestyle = ':',linewidth = 2, alpha = 0.8)
plt.show()
```
