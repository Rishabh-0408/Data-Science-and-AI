```python
# Encoding

from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for i in df.columns:
  if df[i].dtype == 'object':
    df[i] = le.fit_transform(df[i])
```

```python
x = df.drop(columns=['Churn'])
y = df['Churn']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state= 42)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators= 100)
# no of decision tree as 100
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(y_pred,y_test)
```

```python
print(classification_report(y_pred,y_test))
```

```python
df['Churn'].value_counts()
```

```python
 # we can use different n_estimators

n = [50,60,100,200,300,400,500,600,700,800,900,1000]

accuracy = []

for i in n:
  model = RandomForestClassifier(n_estimators=i)
  model.fit(x_train,y_train)
  ans = model.predict(x_test)
  a = accuracy_score(ans,y_test)
  accuracy.append(a)
```

```python
accuracy
```

```python
d = {"Number of Trees" : n, "Accuracy score" : accuracy}
df1 = pd.DataFrame(d)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_estimators':[100,200,300],
    'max_depth':[None, 5,10],
    'min_samples_split': [2,5,10],
    'min_samples_leaf': [1,2,3],
    'criterion': ['gini', 'entropy']
}
```

```python
model2 = RandomForestClassifier(random_state = 0)
grid_search = GridSearchCV(estimator = model2, param_grid = param_grid, cv = 2)
grid_search.fit(x_train,y_train)
```

```python
grid_search.best_estimator_
```

```python
model3 = RandomForestClassifier(criterion='entropy', max_depth=10, min_samples_leaf=2,min_samples_split=5,n_estimators=300,random_state=0)
model3.fit(x_train,y_train)
```

```python
y_pred = model3.predict(x_test)
```

```python
accuracy_score(y_pred,y_test)
```

```python
# 1. n_estimators: This is the number of decision trees
# 2. max_depth: This controls the maximum depth of each dts.
# 3. min_samples_split: This is the minimum number of samples required to split a node.
# 4. min_sample_leaf: This controls the minimum number of samples that each leaf node must contain.
# 5. Criterion: This determines how the tree decides the best split
# gini: uses gini impurity
# entropy: uses information gain
# GridSearchCV test both and picks the one that performs better on your data.
```
