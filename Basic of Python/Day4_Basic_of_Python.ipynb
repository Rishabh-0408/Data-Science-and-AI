Match Case

```python
a = int(input())
b = int(input())
c = int(input("Enter the Choice between 1 to 7: "))
match c:
  case 1:
    print(f"a+b : {a+b}")
  case 2:
    print(f"a-b : {a-b}")
  case 3:
    if(b==0 or a==0):
      print("If any value is zero its multipication is also zero")
    else:
      print(f"a*b : {a*b}")
  case 4:
    if(b==0):
      print("zero Division error")
    else:
      print(f"a/b : {a/b}")
  case 5:
    print(f"a**b : {a**b}")
  case 6:
    print(f"a%b : {a%b}")
  case 7:
    if(b==0):
      print("zero Division error")
    else:
      print(f"a//b : {a//b}")
  case _:
    print("Invalid Choice")
```
