```python
#Q2.Create a class Hotel that allows a customer to book a room, the hotel starts with the avaiable room 5,
#when a booking is made reduce the count,
#the system should show the avaiable rooms and deny the booking if no rooms are avaiable
class Hotel:
  def __init__(self):
    self.avl_room = 5

  def roomBooking(self):
    self.roomBooked = int(input())
    if self.roomBooked <= self.avl_room:
      self.avl_room -= self.roomBooked
      print("Booked room : ", self.roomBooked)
      print("Available room : ", self.avl_room)
    else:
      print("Avaiable room : ", self.avl_room)
      print("No Room Available ")
```

```python
he = Hotel()
he.avl_room
```

```python
he.roomBooking()
```

```python
he.roomBooking()
```

```python
#Q. Create a class BankAccount that has this attributes
# account number
# account holder
# balance = 0
# add methods:
# add deposit
# withdraw
# display bal
class BankAccount:
  def __init__(self):
    self.acc_number = input()
    self.acc_holder = input()
    self.acc_bal = 0
  def addDeposit(self):
    amount = int(input())
    self.acc_bal += amount
    print("The amount has be Deposited")
    print("The Balance Available : ",self.acc_bal)
  def withdraw(self):
    amount = int(input())
    if amount <= self.acc_bal:
      self.acc_bal -= amount
      print("The amount has been Successfully Withdrawn")
    else:
      print("The Amount was Invalid or more the Account balance")
  def Display_bal(self):
    print("The Available Balance Account : ", self.acc_bal)
```

```python
c1 = BankAccount()
```

```python
c1.acc_bal
c1.addDeposit()
c1.Display_bal
c1.withdraw()
c1.Display_bal()
```

```python
#Q.Create a Class rectangle which has attribute lenght and width
# add methods to calculate
# Area
# perimeter
class Rectangle:
  def __init__(self,lenght,width):
    self.lenght = lenght
    self.width = width
  def area(self):
    area = self.lenght * self.width
    print(f"the area of rectangle is {area}")
  def perimeter(self):
    perimeter = 2*(self.lenght+self.width)
    print(f"the perimeter of rectnagle is {perimeter}")
```

```python
t1 = Rectangle(4,5)
t1.area()
```

```python
t1.perimeter()
```
