For Loop -
range() Func- 3 types
1. range with one parameter range(5) -> start = 0 and end = (5-1 = 4)
2. range with two parameter range(2,8) -> start = 2 and end = (8-1 = 7)
3. range with 3 parameter range(2,11,3) -> start = 2, end = 10 and step = 3

(in) operator - checks the range

```python
# range(n) => start = 0 and end = n-1
for i in range (5):
  print(i,end=" ")
```

```python
for i in range (2,8):
  print(i,end=" ")
```

```python
for i in range (2,20,2):
  print(i,end=" ")
```

Q. Finding all the even number in betwen n

```python
n = int(input())
for i in range (2,n,2):
  print(i,end=" ")
```

```python
# negative iteraton
n = 10
for i in range (n,-1,-1):
  print(i,end=" ")
```

Q. Print the factorial of the number

```python
n = int(input())
fact = 1
for i in range (n,0,-1):
  fact *= i
print(fact)
```

Q. Find the sum of n natural numbers

```python
n = int(input())
sum = 0
for i  in range (n,0,-1):
  sum += i
print(sum)
```

Break and  Continue

```python
# Continue
for i in range(1,10):
  if(i==6):
    continue
  print(i,end=" ")
```

```python
# Break
for i in range(1,10):
  if(i==6):
    break
  print(i,end=" ")
```

Q. Print wheather the number is Prime Number or not

```python
n = int(input())
flag = True
for i  in range (2,n):
   if(n%i == 0):
    flag = False
    print("Not a prime number")
    break
if(flag == True):
   print("Its a prime Number")
```

Q. Fibonacci Series

```python
n = int(input())
a = 0
b = 1
for i in range(n):
  print(a)
  c = a + b
  a = b
  b = c
```
