```python
def reverse(num):
  temp = num
  temp_num = 0
  while(temp):
    digit = temp % 10
    temp_num = temp_num * 10 + digit
    temp = temp // 10
  return temp_num

def isPalindrome(num):
  rev_num = reverse(num)
  if(rev_num == num):
    print("Its a palindrome")
  else:
    print("its not a palidrome")

num = int(input())
isPalindrome(num)
```

```python
# Reverse a list without using .reverse() or slicing.
list2 = [1,20,32,4,55,6,7,7,8,9,]
stack = []
for i in list2:
  stack.append(i)

for i in range(len(list2)):
  list2[i] = stack.pop()
print(list2)

left = 0
right = len(list2) - 1
while left < right:
  list2[left],list2[right] = list2[right],list2[left]
  left += 1
  right -= 1

print(list2)
```

```python
# The Classic "FizzBuzz"
'''Problem: Write a program that prints numbers from 1 to 50.
But for multiples of 3, print "Fizz" instead of the number, and for multiples of 5, print "Buzz".
 For numbers which are multiples of both 3 and 5, print "FizzBuzz".'''
for i in range(1,51):
  if(i % 3 == 0 and i % 5 == 0):
      print("FizzBuzz")
  elif(i % 3 == 0):
      print("Fizz")
  elif(i % 5 == 0):
    print("Buzz")
  else:
    print(i)
```

```python
'''2. Palindrome Checker
Problem: Write a function that takes a string as input and returns True
if it is a palindrome (reads the same forwards and backwards) and False otherwise.
Ignore spaces and capitalization.'''
name = "sts"
rev_name = name[::-1]
flag = False
for i in range(len(name)):
  if(name[i] != rev_name[i]):
    flag = True
if(flag == True):
  print("not a valid palindrome")
else:
  print("a valid palindrome")
```
