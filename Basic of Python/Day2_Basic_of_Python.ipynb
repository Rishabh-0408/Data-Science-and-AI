**Convert Seconds into hours and minutes**

```python
sec = int(input("Enter seconds: "))
min = sec // 60
hrs = min // 60
min = min % 60
sec = sec % 60
print("Hours: ",hrs)
print("Minutres: ",min)
print("Sec: ",sec)
```

```python
hrs = int(input())
min = int(input())
sec = int(input())
hrtosec = hrs*3600
mintosec = min*60
totalsec = hrtosec+mintosec+sec
print(totalsec)
```

```python
days = int(input())
year = days // 365
remaining_days = days % 365
day = remaining_days % 30
month = remaining_days // 30
print(f"year : {year} month : {month} day : {day}")
```
