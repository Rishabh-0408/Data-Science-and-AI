```python
from abc import ABC,abstractmethod
class car(ABC):
  @abstractmethod
  def engine_start(self):
    pass
  @abstractmethod
  def engine_stop(self):
    pass
  @abstractmethod
  def headlight_on(self):
    pass
  @abstractmethod
  def headlight_off(self):
    pass
```

```python
class toyota(car):
  def engine_start(self):
    print("engine started")
  def engine_stop(self):
    print("engine stopped")
  def headlight_on(self):
    print("headlight on")
  def headlight_off(self):
    print("headlight off")
```

```python
fortuner = toyota()
fortuner.engine_start()
```

```python
#Q.Create a ATM class with balance 0 and it should be private, make a methods to add deposite, withdraw and check balance
class atm:
  def __init__(self):
    self.__balance = 0
  def deposite(self):
    deposit = int(input())
    self.__balance += deposit
    print("Balance after Deposit: ",self.__balance)
  def withdraw(self):
    withdraw = int(input())
    self.__balance -= withdraw
    print("Balance after withdraw: ",self.__balance)
  def balance(self):
    print("Balance Amount :",self.__balance)
```

```python
rahot = atm()
```

```python
rahot.balance()
```

```python
rahot.deposite()
```

```python
rahot.withdraw()
```

```python
# Q. Create a abstract class make payment with methods like make_payment() impliment with UPI and Card payment
class payment(ABC):
  @abstractmethod
  def make_payment(self):
    pass
class UPI(payment):
  def make_payment(self):
    print("Payment Done Via UPI")
class Card(payment):
  def make_payment(self):
    print("Payment done Via Card")
```

```python
upi = UPI()
upi.make_payment()
```

```python
card = Card()
card.make_payment()
```
