Q. Reverse a number

```python
n = int(input())
rev = 0
while(n!=0):
    rev = rev * 10 + n % 10
    n = n // 10
print(rev)
```

Q. Write code to check a number is armstrong ot not

```python
n = int(input())
count = 0
new = n
while(n != 0):
  count += 1
  n = n // 10
temp = 0
n = new
while(new != 0):
  digit = new % 10
  temp += digit ** count
  new = new // 10
if temp == n:
  print("Armstrong Number")
else:
  print("Not Armstrong Number")
```
