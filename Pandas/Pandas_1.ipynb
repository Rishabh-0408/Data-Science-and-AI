```python
# Dataframe

data = [["krishna",67,"male"],["rishabh",89,"male"],['jancy',90,"female"],['shwan',22,'Female']]
df = pd.DataFrame(data)
df
```

```python
data = [["krishna",67,"male"],["rishabh",89,"male"],['jancy',90,"female"],['shwan',22,'Female']]
df = pd.DataFrame(data,columns=['Name',"marks","gender"])
df
```

```python
# Set_index help to set a column as index
data = {
    "Name": ["Ram","Shyam","Kishan","Bob"],
    "Age": [21,25,45,32],
    "Marks": [90,82,85,54]
}
df = pd.DataFrame(data)
df = df.set_index('Name')
df
```

```python
# loc / iloc

# loc - label based indexing(string)
# iloc - Position based indexing(numerical)
```

```python
data = {
    "Name": ["Ram","Shyam","Kishan","Bob"],
    "Age": [21,25,45,32],
    "Marks": [90,82,85,54]
}
df = pd.DataFrame(data, index=['a','b','c','d'])
df
```

```python
df.loc['b']
```

```python
df.loc["a":"d"]
```

```python
df.loc["b","Age"]
```

```python
df.loc["a":"c","Name":"Age"]
```

```python
df.loc["a":"c","Name":"Marks"]
```

```python
df.loc["a":"c",("Name","Marks")]
```

```python
# Filtering

df.loc[df["Marks"]>80]
# Filter those students who have scored marks higher than 80
```

```python
df.loc[(df["Marks"]>80) & (df["Age"]>30)]
```

```python
df.loc[(df["Marks"]>80) | (df["Age"]>30)]
```

```python
# iloc  = Position (numerical ) based indexing
```

```python
df.iloc[1]  # 0 -a , 1 - b , 2 - c , 3 - d
```

```python
# i can use either loc / iloc to access the records of my dataframe
# we can use iloc in case we have not specified lable or index
```

```python
df.iloc[1:4, 0:2]
```

```python
df.iloc[2:4,0:3] # n-1 rule is applied here
```
