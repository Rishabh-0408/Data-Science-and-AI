```python
for i in df.columns:
  if(df[i].dtype != "object"):
    plt.boxplot(df[i])
    plt.title(i)
    plt.show()
```

```python
# Handling the Outliers

# time -> 8:18
```

```python
outliers_list = ["duration", "campaign", "cons.conf.idx"]
```

```python
# time -> 8:43
```

```python
for i in outliers_list:
  Q1 = df[i].quantile(0.25)
  Q3 = df[i].quantile(0.75)
  IQR = Q3 - Q1
  LB = Q1-1.5*IQR
  UB = Q3+1.5*IQR
  df = df[(df[i]>= LB) & (df[i]<=UB)]
```

```python
for i in df.columns:
  if(df[i].dtype != "object"):
    plt.boxplot(df[i])
    plt.title(i)
    plt.show()
```
